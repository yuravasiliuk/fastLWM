```python



# one episode = a contiguous row slice
ep = 0
s = slice(f["ep_offset"][ep], f["ep_offset"][ep] + ep_len[ep])
pix, act, prop, state = (f[k][s] for k in ("pixels", "action", "proprio", "state"))
assert (f["episode_idx"][s] == ep).all() and (f["step_idx"][s] == np.arange(ep_len[ep])).all()

# frames: 8 evenly spaced
idx = np.linspace(0, len(pix) - 1, 8).astype(int)
fig, ax = plt.subplots(1, 8, figsize=(20, 3))
for a, i in zip(ax, idx):
    a.imshow(pix[i]); a.set_title(f"t={i}"); a.axis("off")

# low-dim signals over time
fig, ax = plt.subplots(3, 1, figsize=(10, 7), sharex=True)
for a, (name, x) in zip(ax, [("action", act), ("proprio", prop), ("state", state)]):
    a.plot(x); a.set_ylabel(name)
ax[-1].set_xlabel("step")

# global action/state distribution on a 100k-row subsample (NaNs = padded actions, see nan_to_num in train.py)
sub = np.sort(np.random.default_rng(0).choice(len(f["action"]), 100_000, replace=False))
a_all = f["action"][sub]
print("NaN action rows:", np.isnan(a_all).any(1).mean())
plt.figure(); plt.hist2d(*a_all[~np.isnan(a_all).any(1)].T, bins=100); plt.title("action dist")
plt.show()
```
